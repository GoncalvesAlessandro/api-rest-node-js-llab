```python
# 1. Instalar a biblioteca de datasets (necessária para baixar a Wikipedia)
!pip install datasets -q

import json
import os
from datasets import load_dataset
from google.colab import drive

drive.mount('/content/drive')
caminho_pasta = '/content/drive/MyDrive/TCC_Base_Dados_POC'

if not os.path.exists(caminho_pasta):
    os.makedirs(caminho_pasta)

print("A descarregar 50.000 artigos da Wikipedia em Português...")
print("(Isto pode demorar cerca de 1 a 3 minutos, dependendo da ligação do Colab)")

# Carregar um recorte (slice) de 50.000 artigos da base oficial da Wikipedia PT
# Usamos a versão "20231101.pt" que é a base estável em português
dataset = load_dataset("wikimedia/wikipedia", "20231101.pt", split="train[:50000]")

print("Descarregamento concluído! A processar e formatar os textos para a POC...")

artigos_wiki = []
for item in dataset:
    texto_limpo = item['text'][:2000].replace('\n', ' ').strip()

    artigos_wiki.append({
        "id": item['id'],
        "titulo": item['title'],
        "texto": texto_limpo
    })

caminho_ficheiro = os.path.join(caminho_pasta, 'dataset_wiki_50k.json')

with open(caminho_ficheiro, 'w', encoding='utf-8') as f:
    json.dump(artigos_wiki, f, ensure_ascii=False, indent=4)

print("----")
print(f"Sucesso! O ficheiro com {len(artigos_wiki)} documentos foi criado.")
print(f"O seu dataset de escala foi guardado em: {caminho_ficheiro}")
```
